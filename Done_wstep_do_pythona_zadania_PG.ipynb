**ZADANIE 1:** Analiza danych sprzedażowych
---------------------------------------
Dane: Miesięczna sprzedaż produktu w PLN

sprzedaz = [12500, 13200, 15800, 14200, 16500, 18200, 17800, 19200,
            20500, 21300, 23500, 25200]

a) Oblicz podstawowe statystyki (średnia, mediana, min, max)

b) Znajdź miesiące ze sprzedażą powyżej średniej

c) Oblicz procentowy wzrost sprzedaży (grudzień vs styczeń)

d) Stwórz wykres liniowy pokazujący trend

e) Dodaj linię średniej na wykresie


````python
## Import bibliotek
import matplotlib.pyplot as plt
import numpy as np

import pandas as pd
import random

plt.rcParams.update({'font.size': 12})
plt.rcParams['figure.figsize'] = (10, 6)
````

````python
sprzedaz = [12500, 13200, 15800, 14200, 16500, 18200, 17800, 19200,
            20500, 21300, 23500, 25200]
print("Sprzedaz (PLN):", sprzedaz)
````

````python
miesiace = ["Styczeń", "Luty", "Marzec", "Kwiecień", "Maj", "Czerwiec", "Lipiec", "Sierpień", "Wrzesień", "Październik", "Listopad", "Grudzień"]
print("Miesiące:", miesiace)
````

````python
def oblicz_statystyki(sprzedaz):
   return {
        "srednia": sum(sprzedaz) / len(sprzedaz),
        "mediana": sorted(sprzedaz)[len(sprzedaz)//2],
        "min": min(sprzedaz),
        "max": max(sprzedaz),
    }
````

````python
wyniki = oblicz_statystyki(sprzedaz)
print("Statystyki sprzedazy:")
for k, v in wyniki.items():
    print(f"  {k}: {v:.2f}")
````

````python
print("Sprzedaż miesięczna:")
for miesiac, kwota in zip(miesiace, sprzedaz):
    print(f"{miesiac}: {kwota} PLN")
````

````python
srednia = wyniki["srednia"]
print(f"Srednia sprzedaz: {srednia:.2f} PLN\n")

print("Sprzedaż miesięczna:")
for miesiac, kwota in zip(miesiace, sprzedaz):
    if kwota > srednia:
        print(f"{miesiac}: {kwota} PLN")
````

````python
print("Procentowy wzrost sprzedaży (grudzień vs styczeń):")

procentowy_wzrost = ((sprzedaz[11] - sprzedaz[0]) / sprzedaz[0]*100)

print(f"{procentowy_wzrost:.2f}%")
````

````python
x = np.arange(len(miesiace))
y = np.array(sprzedaz)

plt.figure(figsize=(15, 4))
plt.plot(x, y, linewidth=8, label='Trend Sprzedaży')
plt.xlabel('Miesiące')
plt.ylabel('Sprzedaż (w PLN)')
plt.title('Sprzedaż miesięczna')
plt.grid(True, alpha=0.5)

````

````python
x = np.arange(len(miesiace))
y = np.array(sprzedaz)

plt.figure(figsize=(10, 4))
plt.plot(x, y, linewidth=1.5, label='Trend Sprzedaży')
plt.xlabel('Miesiące')
plt.ylabel('Sprzedaż (w PLN)')
plt.title('Sprzedaż miesięczna')
plt.grid(True, alpha=0.1)
plt.xticks(x, (x+1), rotation=0)

srednia = np.mean(y)
plt.axhline(y=srednia, color='r', linestyle='--', linewidth=1.5, label=f'Średnia: {srednia:.0f} PLN')


````

c) Oblicz zysk/stratę dla każdej akcji

d) Które akcje przynoszą największy zwrot (w %)?

````python
portfolio = {
    'AAPL': {'liczba': 10, 'cena_zakupu': 150, 'cena_aktualna': 175},
    'GOOGL': {'liczba': 5, 'cena_zakupu': 2800, 'cena_aktualna': 2950},
    'MSFT': {'liczba': 15, 'cena_zakupu': 300, 'cena_aktualna': 320}
}

wartosc_poczatkowa = sum(dane['liczba'] * dane['cena_zakupu'] for dane in portfolio.values())
wartosc_aktualna = sum(dane['liczba'] * dane['cena_aktualna'] for dane in portfolio.values())

print(f"Wartość początkowa: {wartosc_poczatkowa:.2f} USD")
print(f"Wartość aktualna:   {wartosc_aktualna:.2f} USD\n")

print("Zysk/strata dla poszczególnych spółek:\n")


for symbol, dane in portfolio.items():
    koszt = dane['liczba'] * dane['cena_zakupu']
    wartosc = dane['liczba'] * dane['cena_aktualna']
    zysk = wartosc - koszt
    procent = (zysk / koszt) * 100
    zyski[symbol] = procent
    print(f"  {symbol}: zysk {zysk:.2f} USD ({procent:.2f}%)")

    najlepsza_akcja = max(
    portfolio.items(),
    key=lambda item: ((item[1]['cena_aktualna'] / item[1]['cena_zakupu']) - 1)
)


print(f"\n Najlepszy zwrot (%): {najlepsza_akcja[0]} "
      f"({((najlepsza_akcja[1]['cena_aktualna']/najlepsza_akcja[1]['cena_zakupu'])-1)*100:.2f}%)")
````

**ZADANIE 3:** Wizualizacja danych klimatycznych
--------------------------------------------
temp_2022 = [2, 4, 8, 13, 18, 21, 24, 23, 19, 14, 8, 3]


temp_2023 = [3, 5, 9, 14, 19, 22, 25, 24, 20, 15, 9, 4]

miesiace = ['Sty', 'Lut', 'Mar', 'Kwi', 'Maj', 'Cze',
            'Lip', 'Sie', 'Wrz', 'Paź', 'Lis', 'Gru']

a) Stwórz wykres liniowy porównujący temperatury w obu latach

b) Zaznacz najcieplejszy i najzimniejszy miesiąc każdego roku

c) Oblicz średnią różnicę temperatur między latami

d) Dodaj odpowiednie etykiety, legendę i tytuł

````python
# Dane
temp_2022 = [2, 4, 8, 13, 18, 21, 24, 23, 19, 14, 8, 3]
temp_2023 = [3, 5, 9, 14, 19, 22, 25, 24, 20, 15, 9, 4]

miesiace = ['Sty', 'Lut', 'Mar', 'Kwi', 'Maj', 'Cze',
            'Lip', 'Sie', 'Wrz', 'Paź', 'Lis', 'Gru']

x = np.arange(len(miesiace))

# a) Wykres liniowy porównujący temperatury w obu latach
plt.figure(figsize=(10, 5))
plt.plot(x, temp_2022, 'b-o', label='2022', linewidth=2)
plt.plot(x, temp_2023, 'r-o', label='2023', linewidth=2)

# b) Zaznacz najcieplejszy i najzimniejszy miesiąc każdego roku
max_2022 = np.argmax(temp_2022)
min_2022 = np.argmin(temp_2022)
max_2023 = np.argmax(temp_2023)
min_2023 = np.argmin(temp_2023)

plt.scatter(max_2022, temp_2022[max_2022], color='darkblue', s=100, marker='^', label='Najcieplejszy 2022')
plt.scatter(min_2022, temp_2022[min_2022], color='lightblue', s=100, marker='v', label='Najzimniejszy 2022')
plt.scatter(max_2023, temp_2023[max_2023], color='darkred', s=100, marker='^', label='Najcieplejszy 2023')
plt.scatter(min_2023, temp_2023[min_2023], color='pink', s=100, marker='v', label='Najzimniejszy 2023')

# c) Oblicz średnią różnicę temperatur między latami
roznice = [abs(a - b) for a, b in zip(temp_2022, temp_2023)]
srednia_roznica = sum(roznice) / len(roznice)
print(f"Średnia różnica temperatur między 2022 a 2023: {srednia_roznica:.2f}°C")

# d) Etykiety, legenda i tytuł
plt.xticks(x, miesiace)
plt.xlabel('Miesiąc')
plt.ylabel('Temperatura [°C]')
plt.title('Porównanie średnich miesięcznych temperatur: 2022 vs 2023')
plt.legend(loc='best')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

**ZADANIE 4:** Symulacja Monte Carlo
---------------------------------
Zasymuluj 1000 rzutów kostką (6-ścienną):

a) Wygeneruj dane używając np.random.randint(1, 7, 1000)

b) Stwórz histogram pokazujący częstość każdej liczby

c) Oblicz teoretyczne i empiryczne prawdopodobieństwo każdej liczby

d) Czy wyniki są zgodne z teorią? (każda liczba ~16.67%)

````python

# a) Wygeneruj 1000 rzutów kostką (liczby od 1 do 6)
rzuty = np.random.randint(1, 7, 1000)

# b) Histogram częstości każdej liczby
plt.figure(figsize=(8, 5))
plt.hist(rzuty, bins=np.arange(0.5, 7.5, 1), edgecolor='black', rwidth=0.8)
plt.title("Symulacja 1000 rzutów kostką")
plt.xlabel("Wyrzucona liczba")
plt.ylabel("Częstość wystąpień")
plt.xticks(range(1, 7))
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# c) Oblicz teoretyczne i empiryczne prawdopodobieństwa
unikalne, liczby = np.unique(rzuty, return_counts=True)
p_empiryczne = liczby / len(rzuty)
p_teoretyczne = 1 / 6  # dla uczciwej kostki

print("Liczba | Empiryczne Prawdopodobieństwo | Teoretyczne Prawdopodobieństwo")
for liczba, p_emp in zip(unikalne, p_empiryczne):
    print(f"  {liczba:>3}   | {p_emp*100:6.2f}%                     | {p_teoretyczne*100:6.2f}%")

# d) Ocena zgodności z teorią
roznica_srednia = np.mean(abs(p_empiryczne - p_teoretyczne)) * 100
print(f"\nŚrednie odchylenie od teorii: {roznica_srednia:.2f}%")
if roznica_srednia < 2:
    print("✅ Wyniki są zgodne z teorią (różnice mieszczą się w granicach losowości).")
else:
    print("⚠️ Wyniki nieco odbiegają od teorii – można zwiększyć liczbę rzutów.")
````
